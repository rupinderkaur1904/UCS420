```python
# question 2
# a.
arr = np.array([1,2,3,6,4,5])
print(arr[::-1])
# b.(i)
x = np.array([1,2,3,4,5,1,2,1,1,1])
values , counts = np.unique(x,return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x==most_frequent)
print(most_frequent)
print(indices)
```

```python
# b.(ii)
y = np.array([1,1,1,2,3,4,2,4,3,3])
values,counts = np.unique(y,return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y==most_frequent)
print(values)
print(counts)
print(most_frequent)
print(indices)
```

```python
# question 4
Rupinder = np.linspace(10,100,25)
print(Rupinder)
print(f"Dimension is -> {Rupinder.ndim}")
print(f"Shape is -> {Rupinder.shape}") # tells us total elements in each dimention
print(f"total elements -> {Rupinder.size}") # tells us total elements
print(f"Data type -> {Rupinder.dtype}")
print(f"Total no. of bytes consumed -> {Rupinder.nbytes}")
# finding the transpose
Rupinder2 = Rupinder.reshape(5,5)
print(Rupinder2)
print(Rupinder.T) # T attribute does not visually create any difference when used for 1 - D array
```
