```python
import pandas as pd



data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
df = pd.read_csv("employees.csv")
# a.
print("a) Shape of DataFrame:")
print(df.shape)
# b.
print("\nb) Summary of DataFrame:")
df.info()
# c.

print("\nc) Descriptive Statistics:")
print(df.describe())
# d.

print("\nd) First 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))
# e.
# i.
average_salary = df["Salary"].mean()

# ii.
total_bonus = df["Bonus"].sum()

# iii.
youngest_age = df["Age"].min()

# iv.
highest_rating = df["Rating"].max()

print("\ne) Statistics:")
print("i. Average Salary:", average_salary)
print("ii. Total Bonus:", total_bonus)
print("iii. Youngest Employee's Age:", youngest_age)
print("iv. Highest Performance Rating:", highest_rating)
# f.
print("\nf) DataFrame sorted by Salary:")
sorted_df = df.sort_values(by="Salary", ascending=False)
print(sorted_df)
# g.
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(performance_category)

print("\ng) Performance Category:")
print(df[["Name", "Rating", "Performance"]])
# h.
print("\nh) Missing Values:")
print(df.isnull().sum())

# i.
df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\ni) DataFrame after renaming Employee_ID to ID:")
print(df)
# j .

# i.
print("\nj(i) Employees with more than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])

# ii.
print("\nj(ii) Employees belonging to IT department:")
print(df[df["Department"] == "IT"])
# k.

df["Tax"] = df["Salary"] * 0.10

print("\nk) DataFrame with Tax column:")
print(df)
# l .

df.to_csv("employees_modified.csv", index=False)

print("\nl) Modified DataFrame saved as employees_modified.csv")
```
