```python
# question 2
scores = (10, 0, 20, 40, 10, 70, 0, 0)
#part (i)
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_index = scores.index(lowest)
lowest_count = scores.count(lowest)
print("highest score: ", highest)
print("highest score index : ", highest_index)

print("lowest score : ", lowest)
print("lowest score index : ", lowest_index)

print(f"lowest score appeared {lowest_count} times.")
#part (ii)

reversed_tuple_list = list(reversed(scores))
print("reversed_tuple_list:",reversed_tuple_list)

#part (iii)
score = int(input("Enter a score: "))
if score in scores :
  print("First occurence index : ", scores.index(score))
else:
  print("Score is not in the tuple.")
#part (iv)
try:
  scores[0] = 100
except TypeError as e:
  print("Error: ", e)
#part (v)
first_score,second_score,*remaining_scores = scores
print("First score: ",first_score)
print("Second score: ",second_score)
print("remaining scores: ", remaining_scores)
```

```python
# question 3
import random

random.seed(1024170046)
numbers = [random.randint(100,900) for _ in range(100)]
print("Random numbers: ", numbers)
odd_numbers = [num for num in numbers if num%2 != 0]
print("Odd numbers : ",odd_numbers)
even_numbers = [num for num in numbers if num%2 == 0]
print("Even numbers : ", even_numbers)
print("length of even numbers : ", len(even_numbers))
def is_prime(n):
    if n<2:
      return False
    for i in range(2,int(n**0.5)+1):
      if n%i == 0:
        return False
    return True
prime_numbers = [num for num in numbers if is_prime(num)]
print("Prime numbers: ", prime_numbers)
print("length of prime numbers: ",len(prime_numbers))
frequent = max(numbers,key = numbers.count)
frequency = numbers.count(frequent)
print("Most frequent number: ", frequent)
print("Frequency : ", frequency)
```

```python
# question 4
digits = [int(digit) for digit in str(1024170046)[:8]]
A = {digit*7 for digit in digits}
B = {digit*9 for digit in digits}
print("Set A : ", A)
print("Set B : ", B)
# part (vi)
union_set = A.union(B)
print("Union of A and B is : ", union_set)
# part (vii)
intersection_set = A.intersection(B)
print("Intersection of A and B is : ", intersection_set)
# part (viii)
difference1 = A.difference(B)
print("A - B : ",difference1)
difference2 = B.difference(A)
print("B - A: ",difference2)
# difference() gives elements from one set only whereas symmetric difference() gives only the elements that are not common to either set.
#part(ix)
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference: ", symmetric_diff)
# part (x)
print("Is A a subset of B : ", A.issubset(B))
print("Is B a superset of A : ", B.issuperset(B))
#part (xi)
x = int(input("Enter a value from A: "))
A.discard(x)
print("Set A after discard: ", A)
# discard() is safer than remove because if x is absent in A it does not raise an error .
```

```python
#question 5
my_dict = {
    "name" : "Rupinder kaur",
    "roll_no": "1024170046",
    "branch" : "COPC",
    "age" : 19,
    "City" : "Samana"
}
print("Original dictionary : ", my_dict)
# part (i)
my_dict["location"] = my_dict.pop("City")
print("After renaming city: ", my_dict)
# part(ii)
my_dict["cgpa"] = 9.68
print("After adding CGPA : ", my_dict)
# part(iii)
my_dict["age"] = my_dict["age"] + 1
print("After increasing age : ", my_dict)
# part(iv)
pop_dict = my_dict.copy()
removed_branch = pop_dict.pop("branch")
print("Using pop(): ",pop_dict)
print("Value returned by pop(): ", removed_branch)
dict_del = my_dict.copy()
del dict_del["branch"]
print("Using del: ",dict_del)
# pop() removes a key and returns its value whereas del remove only the key and does not return anything.
# part(v)
for k,v in my_dict.items():
  print(k,"->",v)
# part(vi)
if "email" in my_dict:
  print("Email: ",my_dict["email"])
else:
  print("Email is not present in the dictionary")
# part (vii)
friend_dict = {
    "name" : "Ashmeen",
    "roll_no" : "1024170045",
    "branch" : "COPC",
    "age" : 19,
    "city" : "Chandigarh"
}
merged_dict = {**my_dict,**friend_dict}
print("Merged dictionary : ",merged_dict)
# when both dictionaries have same key, the values formt he second dictionary overwrite from the first dictionary
# part(viii)
string_values = {
    key : value
    for key,value in my_dict.items()
    if isinstance(value,str)
}
print("Dictionary containing only string values : ",string_values)
```
