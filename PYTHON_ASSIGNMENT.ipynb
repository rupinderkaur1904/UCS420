**INPUT TWO NUMBERS FROM USER AND COMPARE THEM**

```python
def compare_numbers(num1,num2):
  if num1 > num2:
    print(f"{num1} > {num2}")
  elif num1 < num2:
     print(f"{num1} < {num2}")
  else:
    print(f"{num1} = {num2}")
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
compare_numbers(num1,num2)
```

**INPUT TWO STRINGS AND COMPARE THEM**

```python
def compare_strings(str1,str2):
  if str1 < str2:
    print(f"{str1} < {str2}")
  elif str1 > str2:
    print(f"{str1} > {str2}")
  else:
    print(f"{str1} = {str2}")
str1 = input("Enter the first string: ")
str2 = input("Enter the second string: ")
compare_strings(str1,str2)
```

##8. Strings

Indexing in String

```python
var = 'Hello World!'
print("var --> ",var);
print("var[0] --> ",var[0])
print("var[1:5] --> ",var[1:5])
print("var[:-5] --> ",var[:-5])
```

String length, upper , lower

```python
var = 'Hello World!'
print("string --> ",var)
print("Length --> : ",len(var))
print("Upper --> : ",var.upper())
print("Lower --> : ", var.lower())
```

String strip -> remove unwanted space from the beginning and the end of a string.

```python
var = "Rupinder  kaur  "
print("String -> ",var)
print("Length -> ",len(var))
print("var strip -> ",var.strip())
print("Length of var after striping -> ",len(var.strip()))
```

String Split -> breaking a string into smaller pieces on the basis of a specific character and return them as list

```python
var = " Indian,  Army   "
print("String -> ",var)
print("Length -> ",len(var))
print("var split -> ",var.split())
print("var split -> ",var.split(' '))
print("var split -> ",var.split(','))

# Strip + split
print("var split -> ",var.strip().split(','))
```

Count in string

```python
var = " Indian Army   "
print("String -> ",var)
print("Count of '' -> ",var.count(' '))
print("count of 'a' -> ",var.count('a'))
print("Count of 'an' -> ",var.count('an'))
```

Reverse a string -> just use the pattern string[start:stop:step] with negative indexing

```python
var = "Indian Army"
print("String -> ",var)
print("var[::1] -> ",var[::1])
print("var[::2] -> ",var[::2])
print("var[::-1] -> ",var[::-1])
print("var[::-2] -> ",var[::-2])
var = var[::-1]
print("var after reverse is -> ",var)
```
