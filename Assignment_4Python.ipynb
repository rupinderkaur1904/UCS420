```python
category = ["billing","account","general"]
presonalized_entries = []
for i in two_digits:
  cat = int(i) % 3
  cat_1 = category[cat]
  if cat_1 == 'billing':
    faq = {
        "question" : "How  can i know about my current bill? ",
        "answer" : "By going to bank directly.",
        "keywords" : "Check current bill",
        "category" : cat_1
    }
  elif cat_1 == 'account':
    faq = {
        "question" : "How do I update my registered mobile number. ",
        "answer" : "Go to bank for this please.",
        "keywords" : "Mobile number change.",
        "category" : cat_1
    }
  else:
    faq = {
        "question" : "How are you doing",
        "answer" : "It is good",
        "keywords" : "how you doing",
        "category" : cat_1
    }
  presonalized_entries.append(faq)
```

```python
# question_ 2
def score_query(query,df):
  query_words = query.lower().split()
  result = []
  for index,row in df.iterrows():
    keywords = row['keywords'].lower().split()
    score = 0
    for word in query_words:
      if word in keywords:
        score = score + 1
    if score > 0:
      result.append({
          "question" : row['question'],
          "answer" : row['answer'],
          "score" : score
      })
  result.sort(key=lambda x : x['score'], reverse=True)
  return result

score_query("fee",fixed_entries_df)
```

```python
# question 3
def same_category(category_name,df):
  result = []
  for index,row in df.iterrows():
    if row['category'] == category_name:
      result.append(row['question'])
  return result
```

```python
# question 6
def score_query_tie(query,df):
  query_words = query.lower().split()
  result = []
  for index,row in df.iterrows():
    keywords = row['keywords'].lower().split()
    score = 0
    for word in query_words:
      if word in keywords:
        score = score + 1
    if score > 0:
      result.append({
          "question" : row['question'],
          "answer" : row['answer'],
          "score" : score
      })

  result.sort(key=lambda x : x['score'], reverse=True)

  if len(result) == 0:
    return []

  highest_score = result[0]["score"]
  highest_matches = []

  for entry in result:
    if entry["score"] == highest_score:
      highest_matches.append(entry)

  if len(highest_matches) > 1:
    print("Multiple entries have the highest score:")
    for entry in highest_matches:
      print(entry)
  else:
    print("Highest scoring entry:")
    print(highest_matches[0])

  return result

# Query that produces a tie
score_query_tie("fee",fixed_entries_df)

# Query that does not produce a tie
score_query_tie("password",fixed_entries_df)
```
